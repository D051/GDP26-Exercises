## Teil 1: Fehlersuche


#### Aufgabe 1
**Fehler:** `ArrayIndexOutOfBoundsException`. Die Schleifenbedingung lautet `i <= zahlen.length`. Da Indizes bei 0 beginnen, ist der höchste gültige Index `zahlen.length - 1`.
**Lösung:** Aus `<=` muss `<` gemacht werden.


````java
public class Aufgabe1_Loesung {
    public static void main(String[] args) {
        int[] zahlen = {10, 20, 30, 40};
        for (int i = 0; i < zahlen.length; i++) { // Korrektur hier
            System.out.println(zahlen[i]);
        }
    }
}
````

#### Aufgabe 2
**Fehler:** Referenzverständnis. `b = a` lässt `b` auf dasselbe Array wie `a` zeigen. Durch `b = new int[]{7, 8, 9}` wird `b` jedoch eine *neue* Referenz zugewiesen, `a` bleibt unverändert. Es wird `1` und `7` ausgegeben, nicht zweimal `7`.
**Lösung:** Ersetzt man die Neuzuweisung durch `b[0] = 7`, zeigen `a` und `b` weiterhin auf dasselbe Array. Die Änderung an `b[0]` ist daher auch über `a[0]` sichtbar.


````java
public class Aufgabe2_Loesung {
    public static void main(String[] args) {
        int[] a = {1, 2, 3};
        int[] b = a;
        b[0] = 7; // Korrektur hier: Ändert das geteilte Objekt direkt
        System.out.println(a[0]);
        System.out.println(b[0]);
    }
}
````

#### Aufgabe 3
**Fehler:** `NullPointerException`. Die Array-Variable `werte` zeigt auf `null`. Es wurde kein Array-Objekt im Speicher angelegt, weshalb man keinen Wert zuweisen kann.
**Lösung:** Das Array muss mit `new` initialisiert werden.


````java
public class Aufgabe3_Loesung {
    public static void main(String[] args) {
        int[] werte = new int[1]; // Korrektur hier
        werte[0] = 42;
        System.out.println(werte[0]);
    }
}
````

#### Aufgabe 4
**Fehler:** Es wird keine echte Kopie erstellt. `int[] ergebnis = zahlen;` kopiert nur die Referenz. Das Original-Array wird dadurch ebenfalls verdoppelt!
**Lösung:** Das Array muss z.B. mit `Arrays.copyOf` oder einer eigenen Schleife tatsächlich kopiert werden.


````java
import java.util.Arrays;

public class Aufgabe4_Loesung {
    public static int[] verdoppeln(int[] zahlen) {
        int[] ergebnis = new int[zahlen.length]; // Neues Array erstellen
        for (int i = 0; i < ergebnis.length; i++) {
            ergebnis[i] = zahlen[i] * 2; // Wert aus dem Original lesen und verdoppeln
        }
        return ergebnis;
    }

    public static void main(String[] args) {
        int[] original = {2, 4, 6};
        int[] neu = verdoppeln(original);
        System.out.println(Arrays.toString(original)); // [2, 4, 6]
        System.out.println(Arrays.toString(neu));      // [4, 8, 12]
    }
}
````

#### Aufgabe 5
**Fehler:** Bei Arrays prüft der `==`-Operator, ob beide Variablen auf dasselbe Array-Objekt verweisen. Die beiden Array-Initialisierungen erzeugen verschiedene Objekte, auch wenn ihre Elemente gleich sind.
**Lösung:** Man muss `Arrays.equals()` verwenden.


````java
import java.util.Arrays;

public class Aufgabe5_Loesung {
    public static void main(String[] args) {
        int[] a = {1, 2, 3};
        int[] b = {1, 2, 3};
        
        if (Arrays.equals(a, b)) { // Korrektur hier
            System.out.println("Arrays sind gleich");
        } else {
            System.out.println("Arrays sind verschieden");
        }
    }
}
````

#### Aufgabe 6: Übersetzung nach Java
*Hinweis: In der erfundenen Sprache beginnen die Indizes bei 1 (`a(1)` gibt `1` aus, `b(3)` gibt `6` aus), und die Obergrenze einer `for`-Schleife ist inklusive (siehe `size(a)` in Beispiel 2). In Java müssen wir die Indizes um -1 verschieben (beginnen bei 0) und die Schleife mit `<` statt `<=` schreiben.*


````java
public class Uebersetzung {
    // Beispiel 1
    public static void beispiel1() {
        int[] a = {1, 2, 3};
        int[] b = {4, 5, 6};
        System.out.println(a[0]); // In der erfundenen Sprache: a(1)
        System.out.println(b[2]); // In der erfundenen Sprache: b(3)
        
        int[][] c = {a, b};
        System.out.println(c[1][1]); // In der erfundenen Sprache: c(2,2) -> Array b, zweites Element
    }
    
    // Beispiel 2
    public static int sumUp(int[] a) {
        int s = 0;
        for (int i = 0; i < a.length; i++) {
            s = s + a[i];
        }
        return s;
    }
    
    public static void beispiel2() {
        int[] a = new int[5];
        a[0] = 1;
        a[1] = 2;
        a[2] = 3;
        a[3] = 4;
        a[4] = 5;
        
        int b = sumUp(a);
        System.out.println(b);
    }
    
    public static void main(String[] args) {
        beispiel1();
        beispiel2();
    }
}
````

#### Aufgabe 7: n bis 1 absteigend
Es wird ein Array der Größe $n$ erstellt. Anschließend wird per for-Schleife abwärts iteriert, oder in einer Vorwärts-Schleife umgerechnet. Ein vollständiges Programm, das die Werte ausgibt, ist:


```text
fun generateDescending(in: int n; out: arr(int)) {
    arr(int) res = create(arr(int, n))
    for(i, 1, n) {
        res(i) = n - i + 1
    }
    out(res)
}

fun main(in: null; out: null) {
    arr(int) a = generateDescending(5)
    for(i, 1, size(a)) {
        print(a(i))
    }
}
```


#### Aufgabe 8: Matrix Transponieren
Es wird angenommen, dass die Eingabematrix nicht leer und rechteckig ist. Ein vollständiges Programm ist:


```text
fun transpose(in: arr(arr(int)) m; out: arr(arr(int))) {
    int rows = size(m)
    int cols = size(m(1))
    
    arr(arr(int)) res = create(arr(arr(int), cols))
    for(i, 1, cols) {
        res(i) = create(arr(int, rows))
        for(j, 1, rows) {
            res(i)(j) = m(j)(i)
        }
    }
    out(res)
}

fun main(in: null; out: null) {
    arr(int) zeile1 = [1, 2, 3]
    arr(int) zeile2 = [4, 5, 6]
    arr(arr(int)) m = [zeile1, zeile2]

    arr(arr(int)) t = transpose(m)
    for(i, 1, size(t)) {
        for(j, 1, size(t(i))) {
            print(t(i)(j))
        }
    }
}
```
