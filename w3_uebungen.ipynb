## Teil 3: Fehlersuche

#### Aufgabe 6

Das folgende Programm soll das Maximum von zwei Zahlen bestimmen. Finde den Fehler, erkläre die Fehlerart und behebe ihn.

````java
// Das Programm soll das Maximum von zwei Zahlen bestimmen
class Maximum {

    public static int max(int a, int b) {
        if (a > b) {
            return a;
        }
    }

    public static void main(String[] args) {
        System.out.println(max(4, 9));
    }
}


Maximum.main(new String[0]);
````

#### Aufgabe 7

Das folgende Programm soll eine Zahl verdreifachen. Es lässt sich nicht kompilieren. Finde die Ursache und behebe sie.

````java
// Das Programm soll eine Zahl verdreifachen
class Dreifach {

    public static int dreifach(int zahl) {
        return zahl * 3;
    }

    public static void main(String[] args) {
        System.out.println(dreifach("7"));
    }
}


Dreifach.main(new String[0]);
````

#### Aufgabe 8

Das folgende Programm soll eine Zahl quadrieren und das Resultat ausgeben. Es lässt sich nicht kompilieren. Erkläre, warum, und behebe den Fehler, ohne den Methodennamen zu ändern.

````java
// Das Programm soll eine Zahl quadrieren und ausgeben
class Quadrat {

    public static void quadriere(int x) {
        return x * x;
    }

    public static void main(String[] args) {
        int ergebnis = quadriere(4);
        System.out.println(ergebnis);
    }
}


Quadrat.main(new String[0]);
````
