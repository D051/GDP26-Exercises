### Aufgabe 1 - Gerade [3 Punkte - automatische Tests]

Schreiben Sie eine Methode `isEven` welche überprüft ob die übergebene Zahl gerade ist. Dabei soll für gerade Zahlen `"X ist gerade."` ausgegeben werden und für alle ungeraden Zahlen `"X ist ungerade."`.

Mögliche Ausgaben sind:
- `1 ist ungerade.`
- `2 ist gerade.`
- `7 ist ungerade.`

```java
public class Even {
    public static void isEven(int number) {
        if (number % 2 == 0) {
            System.out.println(number + " ist gerade.");
        } else {
            System.out.println(number + " ist ungerade.");
        }
    }

    public static void main(String[] args) {
        isEven(1);
        isEven(2);
        isEven(7);
        isEven(42);
    }
}
```

### Aufgabe 2 - Maximum [3 Punkte - automatische Tests]

Schreiben Sie eine Methode `max3` welche das Maximum von drei Zahlen zurück gibt. Diese soll für Gleitkommazahlen funktionieren. In dieser und den nachfolgenden Aufgaben, sollen Sie auch immer die main Methode hinzufügen. 

```java
public class MaximumOfThree {
    public static double max3(double a, double b, double c) {
        double max = a;
        if (b > max) {
            max = b;
        }
        if (c > max) {
            max = c;
        }
        return max;
    }

    public static void main(String[] args) {
        System.out.println(max3(1, 2, 3));
        System.out.println(max3(3.5, 2.1, 1.0));
        System.out.println(max3(-1, -5, -3));
    }
}
```

### Aufgabe 3 - Fitze Fatze [4 Punkte - automatische Tests]

Schreiben Sie eine Methode `fitzeFatze`, welche als Parameter eine ganze Zahl entgegen nimmt. 
* Falls die Zahl durch 3 teilbar ist, soll `"fitze"` zurück gegeben werden.
* Falls die Zahl durch 5 teilbar ist, soll `"fatze"` zurück gegeben werden.
* Falls die Zahl durch 3 und 5 teilbar ist, soll `"fitzefatze"` zurück gegeben werden.
* In jedem anderen Fall soll die übergebene Zahl als Test zurück gegeben werden (z.B. für 8 soll "8" zurück gegeben werden). 

*Hinweis: Falls beim lösen dieses Übungsblatt etwas Frustration aufkommen sollte, empfehlen wir Ihnen das zur [Übung passende Lied](https://www.youtube.com/watch?v=neilsnaMI1U) zu hören.*

```java
public class FitzeFatze {
    public static String fitzeFatze(int number) {
        if (number % 3 == 0 && number % 5 == 0) {
            return "fitzefatze";
        } else if (number % 3 == 0) {
            return "fitze";
        } else if (number % 5 == 0) {
            return "fatze";
        } else {
            return Integer.toString(number);
        }
    }

    public static void main(String[] args) {
        System.out.println(fitzeFatze(3));
        System.out.println(fitzeFatze(5));
        System.out.println(fitzeFatze(15));
        System.out.println(fitzeFatze(8));
    }
}
```

- Die Koordinate können Sie als zwei Gleitkommazahlen repräsentieren.
- Der Rand der Form ist immer auf einer ganzen Zahl.
- Sie können wählen ob der Rand zur Form gehört oder nicht.

```java
public class Shape {
    public static boolean isInside(double x, double y) {
        boolean horizontal = x >= -2 && x <= 4 && y >= -1 && y <= 1;
        boolean vertical = x >= 1 && x <= 3 && y >= -2 && y <= 3;
        return horizontal || vertical;
    }

    public static void main(String[] args) {
        System.out.println(isInside(0, 0));
        System.out.println(isInside(0, 1));
        System.out.println(isInside(2, 2));
        System.out.println(isInside(0, 2));
        System.out.println(isInside(3.5, 0));
        System.out.println(isInside(4.5, 0));
        System.out.println(isInside(2, -3));
    }
}
```

### Aufgabe 5 - Dreieck [2 Punkte - automatische Tests]

Für jedes nicht entartete Dreieck gilt: Die Summe zweier beliebiger Seitenlängen muss grösser sein als die dritte Seitenlänge.

Schreiben Sie eine Methode `isTriangle`. Die Methode soll drei Längen übergeben bekommen und zurück geben, ob diese die Seitenlängen eines Dreiecks sein können oder nicht. 

```java
public class Triangle {
    public static boolean isTriangle(double a, double b, double c) {
        return a + b > c && a + c > b && b + c > a;
    }

    public static void main(String[] args) {
        System.out.println(isTriangle(3, 4, 5));
        System.out.println(isTriangle(1, 2, 3));
        System.out.println(isTriangle(2, 2, 2));
    }
}
```

### Aufgabe 6 - Rekursive Summe [3 Punkte - automatische Tests]

Schreiben Sie eine Methode `sumFromZeroTo` welche eine ganze Zahl übergeben bekommt. Die Methode soll alle ganzen Zahlen von 0 bis zur übergebenen Zahl summieren und das Resultat zurückgeben.

Verwenden Sie nicht die Formel $\sum\limits_{n = 0}^Nn = \frac{N \dot (N+1)}{2}$.

```java
public class RecursiveSum {
    public static int sumFromZeroTo(int n) {
        if (n <= 0) {
            return 0;
        }
        return n + sumFromZeroTo(n - 1);
    }

    public static void main(String[] args) {
        System.out.println(sumFromZeroTo(5));
        System.out.println(sumFromZeroTo(10));
    }
}
```

### Aufgabe 8 - Guess my Number [nicht autom. korrigiert]

Wir wollen das Spiel `GuessTheNumber` erweitern, welches Sie auf Blatt 3 in der Aufgabe **Scanner** geschrieben haben. Versuchen Sie folgende 3 Punkte umzusetzten:

- Vergleichen Sie die Zahl welche der Benutzer eingibt mit der gesuchten Zahl und geben Sie als Antwort aus, ob die Zahl gefunden wurde oder nicht.
- Verwenden Sie Rekursion um so lange den Benutzer nach einer neuen Zahl zu fragen, bis die Zahl erraten ist.
- Verwenden Sie Rekursion um den Benutzer so lange nach einer Zahl zu fragen, bis er auch eine Zahl eingegeben hat. Als Test ob eine Zahl eingegeben wurde, können Sie die Methode `hasNextInt()` analog zu `nextLine()` aufrufen. Diese Methode gibt `true` zurück, sobald eine Zahl gelesen werden kann. Die Zahl können Sie dann wie gehabt mit `nextLine()` abfragen.

Als Startpunkt eine mögliche Lösung für die Aufgabe vom letzten Mal.

```java
import java.util.Scanner;

public class GuessMyNumber {
    public static final Scanner SCANNER = new Scanner(System.in);

    public static int askForNumber() {
        System.out.println("Raten Sie eine Zahl zwischen 0 to 10:");
        if (SCANNER.hasNextInt()) {
            return Integer.parseInt(SCANNER.nextLine());
        } else {
            SCANNER.nextLine();
            System.out.println("Das war keine Zahl. Bitte geben Sie eine Zahl ein.");
            return askForNumber();
        }
    }

    public static int generateNumber() {
        double value = Math.random() * 10;
        return (int) Math.floor(value);
    }

    public static void answer(int goal, int guess) {
        if (guess == goal) {
            System.out.println("Sie haben die Zahl " + guess + " gewählt.");
            System.out.println("Ich habe an die Zahl " + goal + " gedacht.");
            System.out.println("Glückwunsch, Sie haben richtig geraten!");
        } else {
            System.out.println("Leider falsch geraten. Versuchen Sie es noch einmal.");
            answer(goal, askForNumber());
        }
    }

    public static void guessMyNumber() {
        int goal = generateNumber();
        answer(goal, askForNumber());
    }

    public static void main(String[] args) {
        guessMyNumber();
    }
}
```

```java
GuessMyNumber.main(new String[0]);
```
