## Teil 1: True-False

#### Aufgabe 1

* Eine void-Methode darf eine return-Anweisung ohne Wert enthalten. **Richtig.** ```return;``` beendet die Methode, ohne einen Wert zurückzugeben.
* Zwei Methoden mit gleichem Namen dürfen gleichzeitig existieren, wenn sie eine unterschiedliche Anzahl Parameter haben. **Richtig.** Dies nennt man Überladung (Overloading).
* Der Rückgabewert einer Methode muss immer einer Variablen zugewiesen werden. **Falsch.** Ein Methodenaufruf darf als eigene Anweisung stehen, der Rückgabewert wird dann einfach ignoriert.

* Eine Methode darf sich selbst aufrufen. **Richtig.** Dies nennt man Rekursion. Wichtig ist eine Abbruchbedingung.
* Eine Methode mit Rückgabetyp ```boolean``` muss mindestens eine return-Anweisung enthalten. **Richtig.** Sonst meldet der Compiler ```missing return statement```.

## Teil 3: Fehlersuche

#### Aufgabe 6

**Fehlerart:** Compilerfehler (```missing return statement```). Wenn ```a > b``` nicht zutrifft, wird keine return-Anweisung ausgeführt. Jeder Programmablauf muss aber einen Wert zurückgeben.

**Behebung:** Ein ```return b;``` nach dem if-Block ergänzen.

````java
// Korrigierte Version
class Maximum {

    public static int max(int a, int b) {
        if (a > b) {
            return a;
        }
        return b;
    }

    public static void main(String[] args) {
        System.out.println(max(4, 9));
    }
}


Maximum.main(new String[0]);
````

#### Aufgabe 7

**Fehlerart:** Compilerfehler (Typfehler beim Methodenaufruf). Die Methode erwartet einen ```int```, übergeben wird aber ein ```String``` (```"7"``` in Anführungszeichen). Java wandelt ```String``` nicht implizit in ```int``` um.

**Behebung:** Ohne Anführungszeichen übergeben: ```dreifach(7)```. Die Ausgabe ist dann ```21```.

````java
// Korrigierte Version
class Dreifach {

    public static int dreifach(int zahl) {
        return zahl * 3;
    }

    public static void main(String[] args) {
        System.out.println(dreifach(7));
    }
}


Dreifach.main(new String[0]);
````

#### Aufgabe 8

**Fehlerart:** Compilerfehler. Die Methode ist als ```void``` deklariert, gibt aber einen Wert zurück. Außerdem kann ein ```void``` nicht einer ```int```-Variablen zugewiesen werden.

**Behebung:** Der Rückgabetyp muss ```int``` sein.

````java
// Korrigierte Version
class Quadrat {

    public static int quadriere(int x) {
        return x * x;
    }

    public static void main(String[] args) {
        int ergebnis = quadriere(4);
        System.out.println(ergebnis);
    }
}


Quadrat.main(new String[0]);
````

## Teil 4: Erfundene Sprachen

#### Aufgabe 9

Die Zahlen der erfundenen Sprache werden in Java als ```double``` umgesetzt. Dadurch gibt Java ```16.0``` statt ```16``` aus, der Wert ist aber derselbe.

Zusatz: ```Print(Square(2.5))``` wird zu ```System.out.println(square(2.5))```. Der Parameter und Rückgabewert sind vom Typ ```double```, weil Gleitkommazahlen verarbeitet werden.

````java
// Übersetzung des Programms
class Square {

    public static double square(double x) {
        double y = x * x;
        return y;
    }

    public static void main(String[] args) {
        System.out.println(square(4));
        System.out.println(square(2.5));
    }
}


Square.main(new String[0]);
````
