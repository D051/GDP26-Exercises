## Teil 1: True-False

#### Aufgabe 1

* In Java muss die Bedingung einer `if`-Anweisung vom Typ `boolean` sein. **Richtig.** Andere Typen wie `int` sind nicht erlaubt.
* Ein `if` kann ohne `else` stehen; der `else`-Zweig ist optional. **Richtig.**
* Bei einer `if`/`else if`/`else`-Kette wird höchstens ein Block ausgeführt. **Richtig.** Der erste zutreffende Zweig gewinnt, die restlichen werden übersprungen.
* `&&` wertet den rechten Operanden immer aus, auch wenn links bereits `false` steht. **Falsch.** Es gilt Kurzschlussauswertung: Ist links schon `false`, wird rechts nicht mehr ausgewertet.
* `!` bindet stärker als `&&`, und `&&` bindet stärker als `||`. **Richtig.** Reihenfolge der Auswertung: `!` vor `&&` vor `||`.
* `!` kann direkt auf einen `int`-Wert angewendet werden. **Falsch.** `!` ist nur für `boolean` definiert.
* Geschachtelte `if`-Anweisungen sind in Java erlaubt. **Richtig.**
* `if (a)` ist gültig, sobald `a` eine von `0` verschiedene Zahl ist. **Falsch.** Die Bedingung muss `boolean` sein; anders als z.B. in C oder Python wird eine Zahl nicht als Wahrheitswert akzeptiert.
* Wenn eine rekursive Methode keine Abbruchbedingung hat, kann sie in einem `StackOverflowError` enden. **Richtig.** Die Aufrufe werden immer weiter auf den Stack gelegt.
* Der Rückgabetyp einer Methode entscheidet, ob sie rekursiv aufgerufen werden darf. **Falsch.** Jede Methode darf sich selbst aufrufen, auch eine `void`-Methode.
* Eine Methode darf eine andere Methode aufrufen, die ihrerseits wieder die erste aufruft (indirekte Rekursion). **Richtig.**
* Jede Schleife lässt sich durch Rekursion ausdrücken und umgekehrt. **Richtig.** Beide sind gleich mächtig; Rekursion braucht meist mehr Speicher.
* Rekursion ist grundsätzlich schneller als eine gleichwertige Schleife. **Falsch.** Jeder Aufruf kostet zusätzlichen Aufwand (Stack); eine Schleife ist meist schneller.

## Teil 2: Boolesche Logik

#### Aufgabe 2

**a)** Auswertung der Ausdrücke:

* `!(6 <= 2)` → `!(false)` = **true**
* `(5 == 5) && (3 > 7)` → `true && false` = **false**
* `(9 % 4 == 1) || (10 / 3 == 3)` → `(1 == 1) || (3 == 3)` = `true || true` = **true**
* `!(true && !false)` → `!(true && true)` = `!true` = **false**
* `(2 < 1) || !(4 == 4)` → `false || !true` = `false || false` = **false**
* `true || false && false` → `true || (false && false)` = `true || false` = **true** (wegen der Präzedenz von `&&` vor `||`)

**b)** Negationen (De Morgan):

* `x > 0 && y > 0` → **`x <= 0 || y <= 0`**
* `a == 5 || b != 3` → **`a != 5 && b == 3`**
* `!(x < 10 && y >= 0)` → **`x >= 10 || y < 0`**

## Teil 3: Fehlersuche

#### Aufgabe 3

**Compilerfehler (`missing return statement`).** Der Zweig `punkte < 50` liefert keinen Wert zurück, obwohl die Methode `String` zurückgeben muss. Behebung: `else`-Zweig ergänzen.

````java
class Noten {

    public static String note(int punkte) {
        if (punkte >= 50) {
            return "bestanden";
        } else {
            return "nicht bestanden";
        }
    }

    public static void main(String[] args) {
        System.out.println(note(30));
        System.out.println(note(70));
    }
}

Noten.main(new String[0]);
````

#### Aufgabe 4

**Logischer Fehler.** `zahl >= 1 || zahl <= 6` ist für *jede* Zahl wahr, denn keine Zahl ist gleichzeitig `< 1` und `> 6`, also trifft immer mindestens eine der Bedingungen zu. Es muss `&&` verwendet werden.

````java
class Wuerfel {

    public static void main(String[] args) {
        int zahl = 9;
        if (zahl >= 1 && zahl <= 6) {
            System.out.println("gültiger Würfelwurf");
        } else {
            System.out.println("kein gültiger Würfelwurf");
        }
    }
}

Wuerfel.main(new String[0]);
````

#### Aufgabe 6

**Compilerfehler (`void cannot be converted to int`).** Die Methode `verdreifache` hat den Rückgabetyp `void` und liefert keinen Wert; das Ergebnis kann deshalb nicht an eine `int`-Variable zugewiesen werden. Behebung: Rückgabetyp auf `int` ändern und den Wert mit `return` zurückgeben.

````java
class Verdreifacher {

    public static int verdreifache(int x) {
        int ergebnis = x * 3;
        System.out.println(ergebnis);
        return ergebnis;
    }

    public static void main(String[] args) {
        int ergebnis = verdreifache(4);
        System.out.println(ergebnis + 1);
    }
}

Verdreifacher.main(new String[0]);
````

#### Aufgabe 7

**Laufzeitfehler (`StackOverflowError`).** Der rekursive Aufruf `produkt(n + 1)` entfernt sich von der Abbruchbedingung `n == 1`, statt sich ihr zu nähern. Dadurch gibt es keine Basis, die Rekursion läuft endlos. Behebung: `produkt(n - 1)`.

````java
class Produkt {

    public static int produkt(int n) {
        if (n == 1) {
            return 1;
        }
        return n * produkt(n - 1);
    }

    public static void main(String[] args) {
        System.out.println(produkt(4));
    }
}

Produkt.main(new String[0]);
````

#### Aufgabe 8

**Logischer Fehler (Dangling Else).** Das `else` gehört zum *inneren* `if` (nächstgelegenes `if`), nicht zum äusseren. Für `zahl = -5` ist das äussere `if` falsch, und das `else` wird gar nicht erreicht – es wird nichts ausgegeben. Behebung: Bedingung zusammenfassen (oder geschweifte Klammern setzen, damit das `else` eindeutig zum äusseren `if` gehört).

````java
class Pruefung {

    public static void main(String[] args) {
        int zahl = -5;
        if (zahl > 0 && zahl > 100) {
            System.out.println("positiv und gross");
        } else {
            System.out.println("nicht gross");
        }
    }
}

Pruefung.main(new String[0]);
````

#### Aufgabe 10

```python
def ist_schaltjahr(jahr):
    return jahr % 4 == 0 and (jahr % 100 != 0 or jahr % 400 == 0)

print(ist_schaltjahr(2024))  # True
print(ist_schaltjahr(1900))  # False
print(ist_schaltjahr(2000))  # True
```

`2024` ist durch `4`, aber nicht durch `100` teilbar → Schaltjahr. `1900` ist durch `4` und `100`, aber nicht durch `400` teilbar → kein Schaltjahr. `2000` ist durch `400` teilbar → Schaltjahr.

#### Aufgabe 11

```python
def summe_bis(n):
    if n <= 0:
        return 0
    return n + summe_bis(n - 1)

print(summe_bis(5))  # 15
```

Der Rekursionsschritt `n + summe_bis(n - 1)` nähert sich mit jedem Aufruf der Abbruchbedingung `n <= 0`.

````java
class Notenrechner {

    public static String note(int punkte) {
        if (punkte >= 90) {
            return "sehr gut";
        } else if (punkte >= 75) {
            return "gut";
        } else if (punkte >= 60) {
            return "genügend";
        } else if (punkte >= 50) {
            return "knapp genügend";
        } else {
            return "nicht bestanden";
        }
    }

    public static void main(String[] args) {
        System.out.println(note(95));
        System.out.println(note(80));
        System.out.println(note(65));
        System.out.println(note(52));
        System.out.println(note(20));
    }
}

Notenrechner.main(new String[0]);
````

````java
class GGT {

    public static int ggT(int a, int b) {
        if (b == 0) {
            return a;
        }
        return ggT(b, a % b);
    }

    public static void main(String[] args) {
        System.out.println(ggT(48, 36));
        System.out.println(ggT(17, 5));
    }
}

GGT.main(new String[0]);
````
