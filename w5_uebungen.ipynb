#### Aufgabe 7

Die Methode soll die Summe der ersten `n` positiven ungeraden Zahlen berechnen (`1 + 3 + 5 + ...`). Für `n = 4` soll das Resultat `16` sein, aber das Programm liefert ein falsches Ergebnis. Finde und behebe den Fehler.

````java
class UngeradeSumme {
    public static int summe(int n) {
        int ergebnis = 0;
        for (int i = 1; i <= n; i++) {
            ergebnis = 2 * i - 1;
        }
        return ergebnis;
    }

    public static void main(String[] args) {
        System.out.println(summe(4));
    }
}

UngeradeSumme.main(new String[0]);
````

#### Aufgabe 9

`istZweierpotenz(n)` soll für eine positive ganze Zahl prüfen, ob sie eine Potenz von `2` ist. Für `n = 1` soll die Methode `true` zurückgeben; für `n = 8` ebenfalls `true`, für `n = 12` dagegen `false`. Das Programm liefert für `8` ein falsches Resultat. Finde und korrigiere den Fehler im Basisfall.

````java
class Zweierpotenz {
    public static boolean istZweierpotenz(int n) {
        if (n == 1) {
            return false;
        }
        if (n % 2 != 0) {
            return false;
        }
        return istZweierpotenz(n / 2);
    }

    public static void main(String[] args) {
        System.out.println(istZweierpotenz(8));
    }
}

Zweierpotenz.main(new String[0]);
````

## Teil 6: Go als Fremdsprache

Go (auch Golang genannt) ist eine echte Programmiersprache. In diesem Teil behandeln wir sie wie eine neue, zunächst unbekannte Sprache und erschliessen die Bedeutung anhand von Beispielen.

Ein paar Hinweise zur Schreibweise:

* `func name(parameter int) int` definiert eine Funktion mit einem `int`-Parameter und einem `int`-Rückgabewert.
* `:=` deklariert eine Variable und weist ihr gleichzeitig einen Wert zu.
* Go verwendet geschweifte Klammern für Blöcke. `for` kann wie in Java mit Zähler, Bedingung und Veränderung oder als `while`-ähnliche Schleife verwendet werden.
* `fmt.Println(...)` gibt Werte aus. `return` gibt einen Wert zurück.
* `package main` und `import "fmt"` stehen am Anfang eines vollständigen Beispielprogramms.

#### Aufgabe 10

Verfolge das Go-Programm von Hand. Welche Zahl wird ausgegeben? Erkläre in eigenen Worten die Aufgaben von `count`, der Schleifenbedingung und der `if`-Bedingung. Übersetze anschliessend die Funktion `anzahlTreffer` nach Java.

```go
package main

import "fmt"

func anzahlTreffer(limit int) int {
    count := 0
    for zahl := 1; zahl <= limit; zahl++ {
        if zahl%4 == 0 && zahl%3 != 0 {
            count++
        }
    }
    return count
}

func main() {
    fmt.Println(anzahlTreffer(20))
}
```

#### Aufgabe 11

```go
package main

func fibonacci(n int) int {
```
