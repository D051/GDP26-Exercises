## Teil 3: Rekursion und Schleifen

#### Aufgabe 3

Der Basisfall ist `n == 0`. Für grössere Werte wird `n²` zur Quadratsumme bis `n - 1` addiert. Für `n = 4` ergibt sich `1 + 4 + 9 + 16 = 30`.

```java
public static int quadrateRec(int n) {
    if (n == 0) {
        return 0;
    }
    return n * n + quadrateRec(n - 1);
}
```

#### Aufgabe 4

Die Schleife summiert die Kehrwerte von `1` bis `n`. `1.0` sorgt dafür, dass die Division als Gleitkommadivision ausgeführt wird.

```java
public static double kehrwerteLoop(int n) {
    double summe = 0.0;
    for (int nenner = 1; nenner <= n; nenner++) {
        summe += 1.0 / nenner;
    }
    return summe;
}
```

Für `n = 4` ist das Ergebnis `1 + 1/2 + 1/3 + 1/4 = 2.083333...`; für `n = 0` wird `0.0` zurückgegeben.

## Teil 4: Prozesse simulieren

#### Aufgabe 5: Sparziel

In jedem Durchlauf wird zuerst der Zins berechnet, dann die Einzahlung addiert und anschliessend die Jahreszahl erhöht. Die Schleife endet, sobald das Ziel erreicht ist. Eine mögliche Lösung ist:

```java
public static int jahreBisZiel(double start, double ziel) {
    double guthaben = start;
    int jahr = 0;
    while (guthaben < ziel) {
        guthaben = guthaben * 1.02;
        guthaben += 80;
        jahr++;
        System.out.printf("Jahr %d: %.2f%n", jahr, guthaben);
    }
    return jahr;
}
```

Mit den angegebenen Werten werden sechs Jahre benötigt. Die Guthaben nach den Jahren sind `590.00`, `681.80`, `775.44`, `870.94`, `968.36` und `1067.73`.

#### Aufgabe 6: Bakterienkultur

Die ersten drei Werte nach einer Stunde sind `50`, `70` und `110`: jeweils die vorherige Anzahl verdoppeln und danach `30` abziehen. Eine Lösung ist:

```java
public static int stundenBisZiel() {
    int bakterien = 40;
    int stunden = 0;
    while (bakterien < 1000) {
        bakterien = bakterien * 2 - 30;
        stunden++;
        System.out.println(bakterien);
    }
    return stunden;
}
```

Die ausgegebenen Werte sind `50`, `70`, `110`, `190`, `350`, `670` und `1310`. Das Ziel wird nach **7 Stunden** erreicht.

## Teil 5: Fehlersuche

#### Aufgabe 7

**Logischer Fehler.** In jedem Durchlauf wird `ergebnis` überschrieben. Stattdessen muss der neue ungerade Wert zum bisherigen Zwischenergebnis addiert werden. Für `n = 4` ist die Summe `1 + 3 + 5 + 7 = 16`; der fehlerhafte Code liefert nur `7`.

```java
public static int summe(int n) {
    int ergebnis = 0;
    for (int i = 1; i <= n; i++) {
        ergebnis += 2 * i - 1;
    }
    return ergebnis;
}
```

#### Aufgabe 8

**Logischer Fehler.** `zahl` wird erhöht, obwohl die Schleife herunterzählen soll. Dadurch bleibt die Bedingung `zahl >= 0` sehr lange wahr; wegen des Überlaufs von `int` endet sie irgendwann, aber nicht wie beabsichtigt. Die Änderung muss `zahl--` sein.

```java
int zahl = 5;
while (zahl >= 0) {
    System.out.println(zahl);
    zahl--;
}
```

Die Ausgabe lautet `5`, `4`, `3`, `2`, `1`, `0`.

#### Aufgabe 9

**Logischer Fehler im Basisfall.** `1` ist selbst eine Zweierpotenz, deshalb muss der Basisfall `true` zurückgeben. Beim Aufruf mit `8` wird rekursiv bis `1` geteilt; der falsche Basiswert macht das Gesamtergebnis `false`.

```java
public static boolean istZweierpotenz(int n) {
    if (n == 1) {
        return true;
    }
    if (n % 2 != 0) {
        return false;
    }
    return istZweierpotenz(n / 2);
}
```

## Teil 6: Go als Fremdsprache

#### Aufgabe 10

Die Bedingung trifft für `4`, `8`, `16` und `20` zu: Diese Zahlen sind durch `4`, aber nicht durch `3` teilbar. Das Programm gibt daher **4** aus. `count` zählt die passenden Zahlen. In Java lautet die Funktion zum Beispiel:

```java
public static int anzahlTreffer(int limit) {
    int count = 0;
    for (int zahl = 1; zahl <= limit; zahl++) {
        if (zahl % 4 == 0 && zahl % 3 != 0) {
            count++;
        }
    }
    return count;
}
```

In Go weist `:=` einer neu deklarierten Variablen einen Startwert zu. `count++` erhöht den Zähler um eins.

#### Aufgabe 11

Die Fibonacci-Folge beginnt mit `0, 1, 1, 2, 3, 5, 8, ...`, also gibt `fibonacci(6)` **8** aus. Die Basisfälle liefern die ersten beiden Werte der Folge; für `n > 1` wird der Wert aus den beiden vorherigen Fibonacci-Zahlen zusammengesetzt. Eine Übersetzung der Funktion nach Java ist:

```java
public static int fibonacci(int n) {
    if (n <= 1) {
        return n;
    }
    return fibonacci(n - 1) + fibonacci(n - 2);
}
```

Der Ansatz ist für kleine `n` gut nachvollziehbar, berechnet aber viele Werte mehrfach.

#### Aufgabe 12

Für jeden möglichen Teiler von `1` bis `n` wird geprüft, ob der Rest `0` ist. Für `12` sind es sechs Zahlen. Ein vollständiges Go-Programm ist:

```go
package main

import "fmt"

func anzahlTeiler(n int) int {
    count := 0
    for teiler := 1; teiler <= n; teiler++ {
        if n%teiler == 0 {
            count++
        }
    }
    return count
}

func main() {
    fmt.Println(anzahlTeiler(12))
}
```

Ausgabe: `6`.
