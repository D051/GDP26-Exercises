## Teil 1: True-False

#### Aufgabe 1

Entscheide für jede Aussage, ob sie richtig oder falsch ist, und begründe deine Entscheidung kurz.

* In Java muss die Bedingung einer `if`-Anweisung vom Typ `boolean` sein.
* Ein `if` kann ohne `else` stehen; der `else`-Zweig ist optional.
* Bei einer `if`/`else if`/`else`-Kette wird höchstens ein Block ausgeführt.
* Der Operator `&&` wertet den rechten Operanden immer aus, auch wenn links bereits `false` steht.
* `!` bindet stärker als `&&`, und `&&` bindet stärker als `||`.
* `!` kann direkt auf einen `int`-Wert angewendet werden.
* Geschachtelte `if`-Anweisungen sind in Java erlaubt.
* `if (a)` ist gültig, sobald `a` eine von `0` verschiedene Zahl ist.
* Der Rückgabetyp einer Methode entscheidet, ob sie rekursiv aufgerufen werden darf.
* Eine Methode darf eine andere Methode aufrufen, die ihrerseits wieder die erste aufruft (indirekte Rekursion).
* Jede Schleife lässt sich durch Rekursion ausdrücken und umgekehrt.
* Rekursion ist grundsätzlich schneller als eine gleichwertige Schleife.

#### Aufgabe 4

````java
// Das Programm soll einen gültigen Würfelwurf prüfen
class Wuerfel {

    public static void main(String[] args) {
        int zahl = 9;
        if (zahl >= 1 || zahl <= 6) {
            System.out.println("gültiger Würfelwurf");
        } else {
            System.out.println("kein gültiger Würfelwurf");
        }
    }
}

Wuerfel.main(new String[0]);
````

#### Aufgabe 7

Das Programm soll das Produkt `n * (n-1) * ... * 1` rekursiv berechnen.

````java
// Das Programm soll n * (n-1) * ... * 1 rekursiv berechnen
class Produkt {

    public static int produkt(int n) {
        if (n == 1) {
            return 1;
        }
        return n * produkt(n + 1);
    }

    public static void main(String[] args) {
        System.out.println(produkt(4));
    }
}

Produkt.main(new String[0]);
````

````java
// Das Programm soll Zahlen prüfen
class Pruefung {

    public static void main(String[] args) {
        int zahl = -5;
        if (zahl > 0)
            if (zahl > 100)
                System.out.println("positiv und gross");
        else
            System.out.println("nicht gross");
    }
}

Pruefung.main(new String[0]);
````
